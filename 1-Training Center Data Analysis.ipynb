```python
#مرات رفع اليد لكل مستوى
avg_hands= df.groupby  ( "Class" )[ "raisedhands" ]  .mean  ()

avg_hands
```

```python
avg_hands.plot ( kind= "bar" ,  color= "skyblue" )
plt.title ( "Average Hand Raise by Level" )
plt.xlabel ( "Level" )
plt.ylabel ( "Average Hand Raise" )
plt.show ()
```
